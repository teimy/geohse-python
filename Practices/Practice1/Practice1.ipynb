### Переменные

- Нужны для хранения информации
- Бывают разные типов:
    - `int` или `integer`: для хранения целых чисел
    - `float`: для хранения нецелых чисел
    - `bool`: для хранения `True` или `False`
    - `string`: для хранения строк (символов или слов)
    - Других типов
- Хранятся в памяти `python` если инициализированы (то есть выполнен код, где определяется переменная)

### Операции над переменными

- Основные операторы:
    - `+`
    - `-`
    - `/`
    - `*`
- Операторы сравнения (как результат возвращают тип `bool`):
    -  `>` — больше, `>=` — больше или равно
    -  `<` — меньше, `<=` — меньше или равно
    -  `==` — проверка равенства (возвращает `True`, если значения одинаковы)
    -  `!=` — не равно
- Логические операторы:
    - `and`
    - `or`
    - `not`
- **Дополнительные операторы**:
    - `x**n` — `x` в степени `n` 
    - `//` — деление нацело
    - `%` — остаток от деления

````python
print('10 + 2 =', 10 + 2)
print('10 - 2 =', 10 - 2)
print('10/2 =', 10/2) # обратите внимание что результат уже не integer
print('10*2 =', 10*2)
print('10 > 2 =', 10 > 2)
print('10 < 2 =', 10 < 2)
print('10 == 2 =', 10 == 2)
print('10 != 2 =', 10 != 2)
print('True and False:', True and False)
print('True or False:', True or False)
print('not True:', not True)
print('2**3 =', 2**3)
print('9 // 4 =', 9 // 4)
print('9 % 4 =', 9 % 4)
````

### Условные операторы:

- оператор `if` позволяет выполнять блок кода при выполнении условия
- следующая конструкция:
    - `if` + условие + :
    - (отступ) Блок кода
    - Блок кода вне отступа выполняется без условий (при отсутствии блока `else`)
- `else`: блок кода, который будет исполняться если условие не выполнено

````python
number = 2 # можно поменять число

if number >= 5:
    print('Число больше или равно пяти')
else:
    print('Число меньше пяти')
````

- `elif`: промежуточный блок, который позволяет определить дополнительное условие

````python
number = 3 # измените число так, чтобы выполнилась ветка `elif`

if number >= 5:
    print('Число больше или равно пяти')
elif number < 0:
    print('Число меньше нуля')
else:
    print('Число меньше пяти, но неотрицательное')
````

trip_total = 31

if trip_total < 5:
    trip_by_foot = trip_total

if (trip_total > 5) & (trip_total <= 20):
    trip_by_foot = 5
    trip_by_bike = trip_total - 5

if trip_total > 20:
    trip_by_foot = 5
    trip_by_bike = 15
    trip_by_train = 100*(trip_total//100)
    trip_by_bus = trip_total%100 - 20
    
CO2 = trip_by_foot*0 + trip_by_bike*0.005 + trip_by_bus*0.01 + trip_by_train*0.005
print('Общее количество CO2:', CO2)
